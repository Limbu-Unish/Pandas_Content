```python
#Importing the pandas module
import pandas as pd
```

```python
#Loading the dataset
dataframe = pd.read_csv('raw_dataset.csv')
```

```python
#setting up the custom number of rows and columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
#Checking the datatypes of all the columns inside the dataset
dataframe.dtypes
```

```python
dataframe.columns = dataframe.columns.str.strip('.').str.replace(r'\xa0', ' ', regex = True)
dataframe.columns
```

```python
#Replacing the unwanted data to a clean data for all the columns.
dataframe['Peak'] = dataframe['Peak'].str.replace(r'\[.*\]', '', regex = True)
dataframe['All Time Peak'] = dataframe['All Time Peak'].str.replace(r'\[.*\]', '', regex= True)
```

```python
#Replacing the unwanted data to a clean data for Actual gross column.

dataframe['Actual gross'] = dataframe['Actual gross'].str.lstrip('$').str.replace(r',|\[.*\]', '', regex = True)
```

```python
#Replacing the unwanted data to a clean data for 'Adjusted gross (in 2022 dollars)' column.
dataframe['Adjusted gross (in 2022 dollars)'] = dataframe['Adjusted gross (in 2022 dollars)'].str.lstrip('$').str.replace(',', '', regex = False)
```

```python
#Replacing the unwanted data to a clean data for 'Your title' column.
dataframe['Tour title'] = dataframe['Tour title'].str.replace(r'†|‡|\[.*\]|\[.*\]|\*', '', regex = True)
```

```python
#Replacing the unwanted data to a clean data for 'Average gross' column.
dataframe['Average gross'] = dataframe['Average gross'].str.replace(r'\$|,', '', regex = True)
dataframe
```

```python
#Replacing the unwanted data to a clean data for 'Ref' column.
dataframe['Ref'] = dataframe['Ref'].replace(r'\[|\]|[a-z]', '', regex = True)
```

```python
#Displaying the dataset
dataframe.head(8)
```

```python
#Changing the data type of multiple columns like peak, all time peak and ref
dataframe['Peak'] = pd.to_numeric(dataframe['Peak'], errors = 'coerce')
dataframe['All Time Peak'] = pd.to_numeric(dataframe['All Time Peak'], errors = 'coerce')
dataframe['Ref'] = pd.to_numeric(dataframe['Ref'], errors = 'coerce')
```

```python
#Replcing the null values in the columns for a relevant data
dataframe['Peak'] = dataframe['Peak'].fillna(dataframe['Peak'].median())
dataframe['All Time Peak'] = dataframe['All Time Peak'].fillna(dataframe['All Time Peak'].median())
dataframe['Ref'] = dataframe['Ref'].fillna(dataframe['Ref'].median())
```

```python
dataframe.rename(columns = {'Ref': 'Reference', 'Actual gross' : 'Actual_Gross'}, inplace = True)
```

```python
dataframe['Actual_Gross'] = pd.to_numeric(dataframe['Actual_Gross'], errors = 'coerce')
```

```python
VAT = 0.13
dataframe['Actual_Gross_with_VAT'] = dataframe['Actual_Gross'] * VAT
```

```python
#Spliting the Years column into two different columns
dataframe[['Starting_Year', 'Ending_Year']] = dataframe['Year(s)'].str.extractall(r'(\d{4})').unstack()
```

```python
#Dropping the Year(s) table after i have splitted it to new columns
dataframe.drop(columns = 'Year(s)', inplace = True)
```
