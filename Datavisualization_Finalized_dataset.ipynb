```python
#importing pandas, numpy and matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Matplotlib basics

```python
#importing the dataset
df = pd.read_csv('Finalized_dataset.csv')
df.columns
```

```python
Artist_name = df['Artist']
Actual_gross = df['Actual_Gross']
```

```python
# creating two arrays x and y
x = np.linspace(0, 5, 11)
y = x * 2
```

```python
#Plotting a linegraph
plt.plot(x, y)
plt.title('Basic example pf title')
plt.xlabel('botton label')
plt.ylabel('right label')
```

```python
#sub plot
plt.subplot(2,2,1)

plt.subplot(2,2,4)
```

```python
#Use of figure in matplotlib
figure = plt.figure(figsize=(10,6), dpi=100)
axes1 = figure.add_axes([0.2, 0.2, 1,1])
axes1.plot(x, y)
axes2 = figure.add_axes([0.1, 0.1, .8,.8])
axes2.plot(x**2,y)
```

```python
#Scatter plot
plt.scatter(x**2, y)
```

```python
#Histogram
data = np.random.randint(100, size = 80)
plt.hist(data)
```

```python
#boxplot
plt.boxplot(data)
```

Practising seaborn

```python
#importing seaborn
import seaborn as sns
```

```python
#distribution plot using sns
fig, axes = plt.subplots(2,2, figsize = (10,10))
sns.histplot(df['All Time Peak'], ax = axes[0,1], kde = True)
sns.histplot(df['Reference'],kde = True, ax = axes[1,0])

```

```python
#Practising joint plot
sns.jointplot(x = df['Actual_Gross'], y = df['Adjusted gross (in 2022 dollars)'], kind = 'reg')
```

```python
#Practising pairplot
#sns.pairplot(df, hue='Artist')
```

```python
#Practising rogplot
sns.rugplot()
```

Categorical Plots

```python
#Count plots
sns.countplot(x = df['Artist'], hue=df['All Time Peak'])
```

```python
#Barplot
import numpy as np
sns.barplot(x = df['Artist'], y = df['Peak'], estimator = np.sum)
```

Matrix pots

```python
#Practising heatmap
data = df[['Rank', 'Peak', 'All Time Peak']]
data_corr = data.corr()

sns.heatmap(data_corr, annot=True)
```

```python
#Practising cluster map
sns.clustermap(data_corr)
```

#Regression plot

```python
sns.lmplot(x = 'All Time Peak', y = 'Actual_Gross', data = df)
```

```python
sns.lmplot(x = 'All Time Peak', y = 'Actual_Gross', data = df, hue= 'Artist', palette='rainbow')
```
